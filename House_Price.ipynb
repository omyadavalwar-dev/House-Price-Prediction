# House Price Prediction

A beginner-level machine learning project using the House Price Prediction dataset.

```python
import pandas as pd
import numpy as np
```

## 1. Load Dataset

Load the CSV file into a Pandas DataFrame.

```python
df = pd.read_csv("HousePricePrediction.csv")
```

```python
df.drop("Id",axis=1,inplace = True)
```

```python
df.drop_duplicates(inplace=True)
```

## 3. Missing Values

Check and handle missing values. Review this section carefully before running it: dropping rows with missing values can remove records intended for prediction.

```python
df["SalePrice"] = df["SalePrice"].dropna()
```

```python
df = df.dropna()
```

```python
df.isna().sum()
```

## 4. Feature Engineering

Create additional features from the existing year columns.

```python
df['HouseAge'] = 2010 - df['YearBuilt']
df['RemodAge'] = 2010 - df['YearRemodAdd']
df['IsRemodeled'] = (df['YearBuilt'] != df['YearRemodAdd']).astype(int)
```

## 5. Encode Categorical Features

Convert selected categorical columns into numerical indicator columns using one-hot encoding.

```python
cols=['MSZoning','LotConfig','BldgType','Exterior1st']
```

```python
df = pd.get_dummies(df, columns=cols, drop_first=True)
```

```python
df
```

```python
bool_cols = df.select_dtypes(include='bool').columns
df[bool_cols] = df[bool_cols].astype(int)
```

## 6. Prepare Cleaned Dataset

Keep a copy of the processed dataset for the next steps.

```python
df_cleaned = df.copy()
```

## 7. Feature Scaling

Standardize selected numerical features.

```python
from sklearn.preprocessing import StandardScaler

# Define numeric columns to scale
num_cols = ['MSSubClass', 'LotArea','OverallCond','YearBuilt','YearRemodAdd','TotalBsmtSF','HouseAge','RemodAge','BsmtFinSF2']

# Initialize and apply Z-score scaling
scaler = StandardScaler()
df_cleaned[num_cols] = scaler.fit_transform(df_cleaned[num_cols])
```

```python
df_cleaned
```

## 8. Train-Test Split

Separate input features (X) and target (y), then split the data into training and testing sets.

```python
# Train-Test Split
from sklearn.model_selection import train_test_split

X = df_cleaned.drop(["SalePrice"],axis = 1)
y = df_cleaned["SalePrice"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
X_train.head()
```

## 9. Model Training

The current code in this notebook uses RandomForestRegressor. The project is still being aligned with the Linear Regression topics covered in the course.

```python
# Train Random Forest Regressor Model
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)
```

## 10. Model Evaluation

Evaluate predictions using R², MAE, RMSE, and MAPE.

```python
# Model Evaluation
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.metrics import mean_absolute_percentage_error
import numpy as np

y_pred = model.predict(X_test)

print("R2 Score:", r2_score(y_test, y_pred))
print("MAE:", mean_absolute_error(y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
print(mean_absolute_percentage_error(y_test, y_pred))
```
